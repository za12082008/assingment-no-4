### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
total_sum = 0

with open("numbers.txt", "r") as file:
    for line in file:
        total_sum += int(line.strip())

print("total sum:", total_sum)
```

### Q13 (Easy)

### Q17

```python
try:
    user_input = int(int("Ek number likhien:"))
    result = 10 / user_input
    print("Result:", result)
except (ZeroDivisionError, ValueError) as e:
    print(f"Error pakda gaya:{e}")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
try:
    num1 = 10
    num2 = 2
    result = num1 / num2
except ZeroDivisionError:
    print("Error: Zero se divide nahi kr sakty!")
else:
    print("koi Error nhi aya! Result hai:", result)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def check_age(age):
    if age < 18:
        raise ValueError("age must br 18 or older")
    return "Access grated!"

try:
    print(check_age(15))
except ValueError as e:
    print(f"Error caught: {e}")
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
file = None
try:
    file = open("sample.txt", "r")
    content = file.read()
    print("file content:\n", content)
except FileNotFoundError:
    print("Error: file nhi mili!")
finally:
    if file and not file.closed:
        file.close()
        print("file successfuly close ho gayi hai.")
```
